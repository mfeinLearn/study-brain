# pandas groupby and indexing (Flatiron Phase 1) — SAMPLE NOTES


Group by two columns and sum another: pass a list of columns to `groupby`, select the column, then aggregate. Use `as_index=False` or `.reset_index()` to get a flat DataFrame back.

```python
import pandas as pd
df = pd.DataFrame({'city': ['NY', 'NY', 'LA'], 'year': [2024, 2024, 2025], 'sales': [10, 5, 7]})
df.groupby(['city', 'year'], as_index=False)['sales'].sum()
```

`loc` selects by label (index names, column names) and its slices include the end. `iloc` selects by integer position and its slices exclude the end, like Python lists.

```python
df.loc[0:1, 'sales']   # rows labelled 0 and 1
df.iloc[0:1, 2]        # only the first row
```
